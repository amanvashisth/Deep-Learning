```python
import numpy as np
import tensorflow as tf
from tensorflow import keras

docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'modi ji ki jai',
		'inquilab zindabad']
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer

# Tokenizer() function is used to convert text 
# into numerical sequences so that a machine learning or deep learning model can process textual data.

tokenizer = Tokenizer(oov_token='<nothing>')

# oov_token is used to handle out-of-vocabulary words. If a word is not present in the training data,
# it will be replaced with the specified token.
```

```python
# The fit_on_texts() function is used to build the vocabulary index based on the input texts.
tokenizer.fit_on_texts(docs)
```

```python
# Assigns a special token to words that are not present in the vocabulary (Out Of Vocabulary words).
tokenizer.word_index
```

```python
# Assigns a special token to words that are not present in the vocabulary (Out Of Vocabulary words).
tokenizer.word_counts
```

```python
# The document_count attribute returns the total number of documents that the tokenizer has seen.
tokenizer.document_count
```

```python
# The texts_to_sequences() function is used to convert the input texts into sequences of integers based on the vocabulary index created by the tokenizer.
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
# The pad_sequences() function is used to ensure that all sequences have the same length by padding shorter sequences with zeros (or another specified value) at the beginning or end.
from tensorflow.keras.preprocessing.sequence import pad_sequences         
```

```python
sequences = pad_sequences(sequences, padding='post')
sequences
```

```python
from keras.datasets import imdb
from keras.models import Sequential
from keras.layers import Flatten, SimpleRNN, Dense
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train[0]
```

```python
len(X_train[0]), len(X_train[1]), len(X_train[2]), len(X_train[3]), len(X_train[4])
```

```python
X_train = pad_sequences(X_train, maxlen=50)
X_test = pad_sequences(X_test, maxlen=50)

# Here we are considering max lenght of 50 for the sequences. 
# This means that all sequences will be padded or truncated to have a length of 50.
```

```python
# Build the model
model = Sequential()

# Add a SimpleRNN layer with 32 units and an input shape of (50, 1)
model.add(SimpleRNN(32, input_shape=(50, 1)))
model.add(Dense(1, activation='sigmoid')) # Add a Dense layer with 1 unit and a sigmoid activation function for binary classification

model.summary()
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X_train, y_train.reshape(-1, 1), epochs=5)
```
