```python
import tensorflow as tf
from tensorflow import keras
```

```python
from keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
img = image.load_img('E:\\Deep Learning\\CNN\\Data Agumentation\\dog.680.jpg', target_size=(200, 200))
```

```python
img = image.img_to_array(img)
```

```python
img.shape
```

```python
img
```

```python
data_gen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)
```

```python
input_batch = img.reshape(1,200,200,3)
```

```python
i = 0 
for output in data_gen.flow(input_batch, batch_size=1, save_to_dir="E:\\Deep Learning\\CNN\\Data Agumentation\\New Agumented Images"):
    i = i + 1
    if i == 10:
        break
```
