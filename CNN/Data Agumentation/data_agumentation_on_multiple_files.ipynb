```python

import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import (
    ImageDataGenerator,
    load_img,
    img_to_array,
    save_img
)

# -------------------------------
# 1. Configure dataset paths
# -------------------------------

INPUT_DIR = "dataset"
OUTPUT_DIR = "augmented_dataset"

# Number of augmented images per original image
AUGMENTATIONS_PER_IMAGE = 5

# Image dimensions
IMAGE_SIZE = (224, 224)

# Supported image formats
VALID_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

# -------------------------------
# 2. Define augmentation methods
# -------------------------------

datagen = ImageDataGenerator(
    rotation_range=25,
    width_shift_range=0.10,
    height_shift_range=0.10,
    shear_range=0.10,
    zoom_range=0.15,
    horizontal_flip=True,
    vertical_flip=False,
    brightness_range=[0.8, 1.2],
    fill_mode="nearest"
)

# -------------------------------
# 3. Process both classes
# -------------------------------

if not os.path.isdir(INPUT_DIR):
    raise FileNotFoundError(
        f"Dataset folder not found: {INPUT_DIR}"
    )

os.makedirs(OUTPUT_DIR, exist_ok=True)

for class_name in os.listdir(INPUT_DIR):

    class_path = os.path.join(INPUT_DIR, class_name)

    # Skip files and non-directory entries
    if not os.path.isdir(class_path):
        continue

    output_class_path = os.path.join(
        OUTPUT_DIR, class_name
    )
    os.makedirs(output_class_path, exist_ok=True)

    image_files = [
        f for f in os.listdir(class_path)
        if f.lower().endswith(VALID_EXTENSIONS)
    ]

    print(f"\nProcessing class: {class_name}")
    print(f"Original images: {len(image_files)}")

    for image_name in image_files:

        image_path = os.path.join(
            class_path, image_name
        )

        try:
            # Load and resize the image
            image = load_img(
                image_path,
                target_size=IMAGE_SIZE,
                color_mode="rgb"
            )

            image_array = img_to_array(image)
            image_array = image_array.reshape(
                (1,) + image_array.shape
            )

            # Create augmented versions
            image_generator = datagen.flow(
                image_array,
                batch_size=1,
                shuffle=False
            )

            base_name = os.path.splitext(
                image_name
            )[0]

            for i in range(AUGMENTATIONS_PER_IMAGE):

                augmented_image = next(
                    image_generator
                )[0]

                output_name = (
                    f"{base_name}_aug_{i + 1}.jpg"
                )

                output_path = os.path.join(
                    output_class_path,
                    output_name
                )

                save_img(
                    output_path,
                    augmented_image,
                    format="jpeg"
                )

            print(f"Augmented: {image_name}")

        except Exception as error:
            print(
                f"Error processing {image_name}: {error}"
            )

print("\nData augmentation completed!")
print(f"Output saved in: {OUTPUT_DIR}")
```
