# Tensors in PyTorch

A tensor is PyTorch's multidimensional array. Tensors can be used for numerical computation on a CPU or, when available, a GPU.

## Import PyTorch

Run this notebook with a Python environment that has PyTorch installed. If the import fails with `ModuleNotFoundError`, install PyTorch into the same environment selected as this notebook's kernel.

```python
import torch

print(torch.__version__)
```

## Create tensors

Tensors can be created from Python values or with helper functions such as `zeros`, `ones`, and `rand`.

```python
scalar = torch.tensor(7)
vector = torch.tensor([1, 2, 3])
matrix = torch.tensor([[1, 2], [3, 4]])
zeros = torch.zeros((2, 3))
ones = torch.ones((2, 3))
random_values = torch.rand((2, 3))

print("Scalar:\n", scalar)
print("Vector:\n", vector)
print("Matrix:\n", matrix)
print("Zeros:\n", zeros)
print("Ones:\n", ones)
print("Random values:\n", random_values)
```

## Inspect tensor properties

The shape describes the dimensions, `dtype` is the element type, and `device` tells you where the tensor is stored.

```python
print("Shape:", matrix.shape)
print("Number of dimensions:", matrix.ndim)
print("Data type:", matrix.dtype)
print("Device:", matrix.device)
```

## Tensor operations

Element-wise arithmetic operates on corresponding elements. Matrix multiplication uses `@` (or `torch.matmul`) and requires compatible shapes.

```python
a = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
b = torch.tensor([[5.0, 6.0], [7.0, 8.0]])

print("Addition:\n", a + b)
print("Element-wise multiplication:\n", a * b)
print("Matrix multiplication:\n", a @ b)
```

## Index and reshape

Indexing selects values from a tensor. Reshaping changes its dimensions without changing the number of elements.

```python
values = torch.arange(1, 7)
print("First element:", values[0].item())
print("Reshaped to 2 rows and 3 columns:\n", values.reshape(2, 3))
```
